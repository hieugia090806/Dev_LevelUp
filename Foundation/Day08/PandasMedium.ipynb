```python
import numpy as np
import pandas as pd
#-- Load the dataset. --#
df = pd.read_csv("student_grade_dirty.csv")

#-- Clean 'Age' column: Replace string text, convert negative values to NaN, and fill with median. --#
df["Age"] = df["Age"].replace("twenty", 20)
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Age"] = df["Age"].apply(lambda x: np.nan if x < 0 else x)
df["Age"] = df["Age"].fillna(df["Age"].median())

#-- Clean 'Grade' column: Convert 'N/A' to NaN, treat outliers (>100) as NaN, and fill with mean. --#
df["Grade"] = pd.to_numeric(df["Grade"], errors="coerce")
df["Grade"] = df["Grade"].apply(lambda x: np.nan if x > 100 else x)
df["Grade"] = df["Grade"].fillna(df["Grade"].mean())

#-- Clean 'Absences' column: Fill missing values with 0. --#
df["Absences"] = pd.to_numeric(df["Absences"], errors="coerce").fillna(0)

print("--- TRANSFORMED DATASET ---")
print(df[["StudentID", "FirstName", "Age", "Grade", "Absences"]].head(10))
```

```python
# 1. Strip leading and trailing whitespaces from names
df["FirstName"] = df["FirstName"].str.strip()

# 2. Standardize 'Gender' column to Capitalized format (e.g., Male, Female)
df["Gender"] = df["Gender"].str.strip().str.capitalize()

# 3. Handle missing values in 'CurriculumActivities' by replacing NaN with 'None'
df["CurriculumActivities"] = df["CurriculumActivities"].fillna("None")

print("--- CLEANED STRING COLUMNS ---")
print(df[["FirstName", "Gender", "CurriculumActivities"]].head(10))
```

```python
# Group by Major and calculate summary statistics
df_summary = (
    df.groupby("Major")
    .agg(
        Student_Count=("StudentID", "count"),
        Average_Grade=("Grade", "mean"),
        Total_Absences=("Absences", "sum"),
    )
    .reset_index()
)

print("--- AGGREGATED METRICS BY MAJOR ---")
print(df_summary)
```
