```python
#-- Filter data. --#
age_filter = student_grade_data[student_grade_data["age"] >= 20]
age_filter.to_csv('age_filter.csv', encoding='utf-8-sig')
```

```python
gender_filter = student_grade_data[student_grade_data["gender"] == "Female"]
gender_filter.to_csv('female_gender_filter.csv', encoding='utf-8-sig')
```

```python
major_and_curriculum_activity = student_grade_data[(student_grade_data["major"] == "AI") & (student_grade_data["curriculum_activities"] == "Volunteer")]
major_and_curriculum_activity.to_csv('major_and_curriculum_activity.csv', encoding='utf-8-sig')
```
