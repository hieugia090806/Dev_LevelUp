```python
#-- Panda series. --#
#-- Import Pandas Library. --#
import pandas as pd
#-- Create Series. --#
temperatures = pd.Series([32, 35, 33, 34], index=["Mon", "Tues", "Wed", "Thurs"])
#-- Print the Series (Work like key-value in dictionary). --# 
print(temperatures)
```
