```python
#-- Import Python Linraries and Data. --#
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
student_data = pd.read_csv('student_grade.csv')
```

```python
#-- Basically. --#
sns.set_theme(style="whitegrid")
#-- Count which major as most students. --#
plt.figure(figsize=(10, 5))
sns.countplot(data=student_data, x="major", palette="Set2")
plt.title("Number of students based on major")
plt.xticks(rotation=15) 
plt.show()
```

```python
sns.barplot(data=student_data, x="major", y="age", errorbar=None, palette="Blues_d")
plt.title("Average age of student according to major")
plt.xticks(rotation=15)
plt.show()
```

```python
sns.boxplot(data=student_data, x="gender", y="absences", palette="pastel")
plt.title("Compare Absences Day based on major")
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=student_data, x="major", y="absences", hue="tutoring", palette="Set1")
plt.title("Relationship between Major, Tutoring, and Absences Day")
plt.show()
```

```python
pivot_table = pd.crosstab(student_data["parental_support"], student_data["tutoring"])
sns.heatmap(pivot_table, annot=True, cmap="YlGnBu", fmt="d")
plt.title("Ma trận phân bổ: Sự hỗ trợ của phụ huynh vs Việc học thêm")
plt.show()
```

```python
g = sns.catplot(
    data=student_data, 
    x="gender", 
    y="absences", 
    hue="tutoring", 
    col="parental_support", 
    kind="bar", 
    palette="muted",
    ci=None
)
g.set_titles("Support from parents: {col_name}")
plt.show()
```
