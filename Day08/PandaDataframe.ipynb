```python
#-- Using dictionary data. --#
data = {
    'name': ['Nam Tran', 'Huy Nguyen', 'Minh Huy', 'Lam Le', 'Dat Nguyen'],
    'city': ['HCM city', 'Ha Noi city', 'Ha Noi city', 'Da Nang city', 'HCM city'],
    'age': [18, 19, 23, 17, 22],
    'py-core': [82.0, 79.0, 89.0, 96.0, 94.0]
}
row_labels = [1, 2, 3, 4, 5]
#-- Create Pandas Dataframe. --#
df = pd.DataFrame(data=data, index=row_labels)
#-- Print df. --#
print(df) #-- df still right. --#
```

```python
import pandas as pd
a = [1, 7, 2]
myvar = pd.Series(a)
print(myvar)
print(myvar[0])
#-- Create lables. --#
myvar = pd.Series(a, index = ["x", "y", "z"])
print(myvar)
```

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
#load data into a DataFrame object:
df = pd.DataFrame(data)
print(df) 
#-- Locate Row. --#
print('-' * 20)
print(df.loc[0]) #-- Print Row 0. --#
print('-' * 20)
print(df.loc[[0, 1]]) #-- Print Row 0 and 1. --#
#-- Name Indexes. --#
print('-' * 20)
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(df) 
#-- Locate Name Indexes. --#
print('-' * 20)
print(df.loc["day2"])
```
