```python
#-- Showcase the dataset. --#
#-- Import Pandas Library. --#
import pandas as pd
#-- Load dirty data. --#
student_grade_dirty_data = pd.read_csv('student_grade_dirty.csv')
#-- Print the data. --#
print(student_grade_dirty_data.to_string())
```

```python
#-- Checking data information. --
print(student_grade_dirty_data.info())
```

```python
#-- Checking and count Missing values in each field. --#
print(student_grade_dirty_data.isnull().sum())
```

```python
clean_df = student_grade_dirty_data.dropna()
print(clean_df.shape)
```

```python
#-- Replace mssing values. --#
student_grade_dirty_data["Grade"] = student_grade_dirty_data["Grade"].fillna(student_grade_dirty_data["Grade"].mean())
```

```python
student_grade_dirty_data["Absences"] = student_grade_dirty_data["Absences"].fillna(student_grade_dirty_data["Absences"].mean())
```

```python
student_grade_dirty_data["CurriculumActivities"] = student_grade_dirty_data["CurriculumActivities"].fillna("Unknown")
```

```python
student_grade_dirty_data["Age"] = pd.to_numeric(student_grade_dirty_data["Age"], errors="coerce")
student_grade_dirty_data["Age"] = student_grade_dirty_data["Age"].fillna(student_grade_dirty_data["Age"].median())
```

```python
#-- Check again. --#
print(student_grade_dirty_data.isnull().sum())
```

```python
#-- Checking Duplicate Data. --#
print(student_grade_dirty_data.duplicated().sum())
```

```python
print(student_grade_dirty_data[student_grade_dirty_data.duplicated(keep=False)])
```

```python
print(f"Before: {len(student_grade_dirty_data)}")
cleand_student_grade = student_grade_dirty_data.drop_duplicates()
print(f"After: {len(cleand_student_grade)}")
```
